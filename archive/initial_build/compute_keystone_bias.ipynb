```python
"""Phase-1 validation: smile(+keystone) truth via gert dispersion, retrieve with
dispersion in the state, compare nonlinear vs linear bias for XCO2/XCH4/XCO."""
import sys, time, numpy as np
from pathlib import Path
sys.path.insert(0, "/scratch/scrowel3_lab/geocarb_simulator")
GERT = Path("/scratch/scrowel3_lab/gert")
sys.path.insert(0, str(GERT))

import geosat_geometry as gg
from geocarb_gert import (build_geocarb_instrument, sample_geometries,
                          reference_atmosphere, albedo_for)
import gert
from gert.forward_model import ForwardModel
from gert.rt_solver import SingleScatterSolver
from gert.retrieval import StateVector, GERTRetrieval
from gert.instrument import FlatSNR
import matplotlib.pyplot as plt
```

```python
# ---- scene / instrument / geometry ----
block = gg.geocarb_demo(verbose=False)['blocks'][0]
row, col, geo = sample_geometries(block, n=1, seed=0)[0]
inst  = build_geocarb_instrument()
atm   = reference_atmosphere()
absco = gert.ABSCOTable.load_all(str(GERT/'input/absco/absco.h5'))
solar = gert.SolarSpectrum.load(str(GERT/'input/solar/solar.h5'))
nb    = len(inst.windows)
albedo = albedo_for(inst, 'desert')          # per-band Lambertian albedo
GASES = ['co2', 'ch4', 'co', 'h2o']
KEY_FRAC = 20.0 / 1024.0                      # 20-px slit-image growth on 1024-px slit
SMILE_PX = 2.0                                # per-band smile (placeholder, TBD)

fm = ForwardModel(atm, absco, inst, geo, solver=SingleScatterSolver(), solar_spectrum=solar)
```

```python
def smile_disp_coeffs(eta, keystone_on, fit_order=4):
    """Per-band dispersion coeffs [a0..] (cm-1) for the smile(+keystone) shift at eta."""
    disp = {}
    for b, w in enumerate(inst.windows):
        nu   = np.asarray(w.wn_instrument, float)
        span = w.wn_max - w.wn_min
        f    = (nu - w.wn_min) / span
        dnu_det = span / 1023.0               # detector pixel = band / 1024 px
        stretch2 = (1.0 + KEY_FRAC * f) ** 2 if keystone_on else 1.0
        dv   = -(SMILE_PX * dnu_det) * f * eta**2 / stretch2      # cm-1 per channel
        u    = w.dispersion_unit_coord()
        disp[b] = np.polyfit(u, dv, fit_order)[::-1]              # -> [a0,a1,...]
    return disp

# nominal (undistorted) truth + noise model
res0   = fm.run(albedo=list(albedo), albedo_slope=[0.0]*nb)
y_nom  = res0.y
sigma  = np.maximum(FlatSNR(300.).sigma(res0.R_band, inst.windows), 1e-30)
Sy_inv = np.diag(1.0 / sigma**2)
xtrue  = {g: atm.column_xgas(g) * (1e6 if g == 'co2' else 1e9) for g in ('co2','ch4','co')}
print("truth Xgas:", {g: round(v,2) for g,v in xtrue.items()}, "| n_y =", y_nom.size)

def bias_case(eta, keystone_on, order):
    disp   = smile_disp_coeffs(eta, keystone_on)
    y_dist = fm.run(albedo=list(albedo), albedo_slope=[0.0]*nb, dispersion=disp).y
    sv = StateVector.gas_scaling(prior_albedo=albedo, prior_albedo_slope=np.zeros(nb),
                                 gases=GASES, include_dispersion=True,
                                 dispersion_order=order, dispersion_uncert=2.0)
    # NOTE: dx_norm (not the default chisq<1.01) convergence criterion.
    # chisq-based stopping can declare convergence after a single GN step
    # whenever chisq happens to cross 1.01, even with the step still large
    # (Delta chisq >> 0) -- this produced a spurious discontinuity in the
    # bias(eta) curve around eta=0.70/0.71 where one side took 1 step and
    # the other took 2. dx_norm stops on actual step size instead.
    ret = GERTRetrieval(fm, y_dist, Sy_inv, sv, prior_albedo=albedo,
                        prior_albedo_slope=np.zeros(nb), analytical_jacobians=True,
                        max_iter=12, verbose=False,
                        convergence_criterion='dx_norm', dx_tol=0.01)
    assert ret.analytical_jacobians, "expected analytical jacobians for SingleScatter"
    names = sv.names
    def xgas_bias(scale_vec):
        return {g: xtrue[g] * (scale_vec[names.index(g+'_scale')] - 1.0) for g in ('co2','ch4','co')}
    # nonlinear
    res = ret.run()
    nl = xgas_bias(res.x_ret); nl['_conv'] = res.converged; nl['_chi2'] = float(res.chisq_reduced)
    # linear: dx = G dy at prior
    x0 = sv.prior; y0 = ret._forward(x0); K, _ = ret._jacobian_mixed(x0, y0)
    mask = sv.active_mask; Sa = sv.Sa
    Ka = K[:, mask]; Sai = np.linalg.inv(Sa[np.ix_(mask, mask)])
    KtSi = Ka.T @ Sy_inv; Shat = np.linalg.inv(KtSi @ Ka + Sai)
    dx = np.zeros(len(sv.names)); dx[mask] = Shat @ (KtSi @ (y_dist - y0))
    lin = xgas_bias(x0 + dx)
    return nl, lin, res, y_dist
```

```python
t=time.time()
slit_pos = np.arange(0.0,0.91,0.05)
dct = {}
for eta in slit_pos:
    nl, lin, res, y_dist = bias_case(eta, True, 2)
    dct[np.round(eta, 2)] = (nl, lin, res, y_dist)
    print("\n[validation] eta=%.2f, smile+keystone, order-2  (%.1fs)"%(eta, time.time()-t))
    print("  nonlinear bias:", {g: round(nl[g],4) for g in ('co2','ch4','co')}, "conv=%s chi2=%.3f"%(nl['_conv'],nl['_chi2']))
    print("  linear    bias:", {g: round(lin[g],4) for g in ('co2','ch4','co')})
```

```python
# Retrieved dispersion coefficients vs slit position (coarse sweep, order-2),
# compared to the analytic order-2 fit of the true smile+keystone dispersion
# at the same eta -- checks that the retrieval is actually tracking the
# right per-band dispersion curve rather than just zeroing the gas bias
# via some other route.
sv_ref = StateVector.gas_scaling(prior_albedo=albedo, prior_albedo_slope=np.zeros(nb),
                                 gases=GASES, include_dispersion=True,
                                 dispersion_order=2, dispersion_uncert=2.0)
names    = sv_ref.names
disp_idx = {name: names.index(name) for name in names if name.startswith('disp_')}

etas_c = sorted(dct)
ncoef  = 3   # a0, a1, a2 for order-2
fig, axes = plt.subplots(nb, 1, sharex=True, figsize=(7, 2.2 * nb))
for b in range(nb):
    for k in range(ncoef):
        name      = f'disp_a{k}_{b}'
        ret_vals  = [dct[e][2].x_ret[disp_idx[name]] for e in etas_c]
        true_vals = [smile_disp_coeffs(e, True, fit_order=2)[b][k] for e in etas_c]
        axes[b].plot(etas_c, ret_vals,  'o-', color=f'C{k}',
                     label=f'a{k} retrieved' if b == 0 else None)
        axes[b].plot(etas_c, true_vals, '--', color=f'C{k}',
                     label=f'a{k} true (order-2 fit)' if b == 0 else None)
    axes[b].set_ylabel(f'band {b}\n$cm^{{-1}}$')
axes[0].legend(fontsize=8, ncol=3, loc='upper left')
axes[-1].set_xlabel('eta (slit position)')
fig.suptitle('Retrieved vs true dispersion coefficients across the slit')
plt.tight_layout()
```

## Phase-2-lite: does residual along-slit radiometric non-uniformity reintroduce bias?

Everything above uses a *perfectly* uniform scene, where keystone is exactly null
(plan Sec. 2a) and dispersion order-2 fully absorbs smile+coupling. Real ground-test
data (upward-looking, slit filled with the same light) is unlikely to be perfectly
uniform after flat-fielding -- residual slit-thickness brightness variation, diffuser
non-uniformity, and radiometric cal artifacts all leave a small along-slit albedo
gradient behind. Unlike a uniform brightness offset (which the retrieval's own
per-band albedo state trivially absorbs), a gradient across the slit gets seen
*differently by different wavelengths within a band*, because keystone maps a
detector row's channel at spectral fraction `f` to a true slit position
`eta_obj(f) = eta / stretch(f)`, `stretch(f) = 1 + KEY_FRAC*f`. That is the actual
mechanism behind Phase 2's "irreducible keystone-heterogeneity bias" -- and it is not
absorbable by floating dispersion, no matter how loose the prior.

We use gert's native `SurfaceBasis` / `surface_basis=` mechanism (an exact,
already-validated per-wavenumber albedo correction -- see `gert/surface_basis.py`)
to build the truth with `rho(nu) = albedo[b] * (1 + amp * eta_obj(nu))`, a linear
along-slit gradient of fractional amplitude `amp`, then retrieve with the *same*
scalar-albedo + order-2-dispersion state as before (nothing new on the retrieval
side) and see how much bias comes back at a few plausible `amp` levels.

```python
def keystone_albedo_basis(eta, amp, keystone_on=True):
    """Per-band SurfaceBasis for a linear along-slit albedo gradient of
    fractional peak amplitude `amp`, sampled at the keystone-mapped true
    slit position eta_obj(f) = eta / stretch(f) seen by each channel."""
    basis = {}
    for b, w in enumerate(inst.windows):
        nu      = np.asarray(w.wn_instrument, float)
        span    = w.wn_max - w.wn_min
        f       = (nu - w.wn_min) / span
        stretch = (1.0 + KEY_FRAC * f) if keystone_on else 1.0
        eta_obj = eta / stretch
        B = np.column_stack([np.ones_like(nu), eta_obj])
        c = np.array([albedo[b], amp * albedo[b]])
        basis[b] = (gert.SurfaceBasis.custom(B, nu, label=f'keygrad_b{b}'), c)
    return basis

def bias_case_grad(eta, amp, order, keystone_on=True):
    disp   = smile_disp_coeffs(eta, keystone_on)
    sbasis = keystone_albedo_basis(eta, amp, keystone_on)
    y_dist = fm.run(albedo=list(albedo), albedo_slope=[0.0]*nb,
                    dispersion=disp, surface_basis=sbasis).y
    sv = StateVector.gas_scaling(prior_albedo=albedo, prior_albedo_slope=np.zeros(nb),
                                 gases=GASES, include_dispersion=True,
                                 dispersion_order=order, dispersion_uncert=2.0)
    ret = GERTRetrieval(fm, y_dist, Sy_inv, sv, prior_albedo=albedo,
                        prior_albedo_slope=np.zeros(nb), analytical_jacobians=True,
                        max_iter=12, verbose=False,
                        convergence_criterion='dx_norm', dx_tol=0.01)
    res = ret.run()
    names = sv.names
    nl = {g: xtrue[g] * (res.x_ret[names.index(g+'_scale')] - 1.0) for g in ('co2','ch4','co')}
    nl['_conv'] = res.converged; nl['_chi2'] = float(res.chisq_reduced)
    return nl, res, y_dist

# Sanity check: amp=0 must reproduce the uniform-scene (near-zero-bias) result
# exactly -- the surface_basis correction should vanish identically.
nl0, res0_chk, y0_chk = bias_case_grad(0.9, 0.0, 2)
print("amp=0 check @ eta=0.9: co2 bias = %.6f (should match the ~0 uniform-scene result)" % nl0['co2'])
```

```python
# Sweep along-slit gradient amplitude x slit position, order-2 dispersion retrieval
amps = [0.0, 0.001, 0.005, 0.01, 0.02, 0.05]     # 0%, 0.1%, 0.5%, 1%, 2%, 5% fractional
etas_g = [0.0, 0.3, 0.5, 0.7, 0.9]
grad_dct = {}
t0 = time.time()
for amp in amps:
    for eta in etas_g:
        nl, res, y_dist = bias_case_grad(eta, amp, 2)
        grad_dct[(amp, eta)] = nl
        print("amp=%.3f  eta=%.1f  chi2=%.4f  conv=%s  co2=%.4f  ch4=%.4f  co=%.4f  (%.1fs)" %
              (amp, eta, nl['_chi2'], nl['_conv'], nl['co2'], nl['ch4'], nl['co'], time.time()-t0))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharex=True)
for ax, g, ylabel in zip(axes, ('co2', 'ch4', 'co'), ('CO2 bias (ppm)', 'CH4 bias (ppb)', 'CO bias (ppb)')):
    for amp in amps:
        vals = [grad_dct[(amp, eta)][g] for eta in etas_g]
        ax.plot(etas_g, vals, 'o-', label=f'{amp*100:.1f}%')
    ax.set_xlabel('eta (slit position)')
    ax.set_ylabel(ylabel)
axes[0].legend(title='along-slit\ngradient amp', fontsize=8)
fig.suptitle('Bias vs slit position, at plausible along-slit radiometric non-uniformity levels')
plt.tight_layout()
```
